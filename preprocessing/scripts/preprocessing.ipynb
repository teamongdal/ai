```python
import pandas as pd
import numpy as np
```

```python
df_raw = pd.read_csv('../data/articles.csv')

df = df_raw
df.head()
```

```python
df.info()
```

```python
len(df['product_group_name'].unique()) # product_group_name 19개의 그룹
df['product_group_name'].value_counts()
```

```python
filtered_df = df[df['product_group_name'].isin([
    'Garment Upper body', 'Garment Lower body', 'Garment Full body'])]

len(filtered_df)
```

```python
len(filtered_df['product_type_name'].unique()) # product_type_name 26개
```

```python
filtered_df['product_type_name'].value_counts()
```

```python
filtered_df['product_type_name'].unique()
```

```python
# 제거할 product_type_name 목록
exclude_types = ['Leggings/Tights', 'Trousers', 'Bodysuit', 'Skirt', 'Shorts', 'Outdoor trousers']

# 제외 조건 적용
filtered_df = filtered_df[~filtered_df['product_type_name'].isin(exclude_types)]

filtered_df['product_type_name'].value_counts()
```

```python
len(filtered_df)
```

```python
filtered_df.to_csv('../data/hnm_fin.csv', index=False)
```
